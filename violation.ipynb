```python
class Violation:
  def __init__(self, driver_id,violation_type, fine_amount,speed_over_limit, numbere_of_violation, payment_status, violation_month):
    self.driver_id = driver_id
    self.violation_type = violation_type
    self.fine_amount = fine_amount
    self.speed_over_limit = speed_over_limit
    self.number_of_violation = self.number_of_violation
    self.payment_status = payment_status
    self.violation_month = violation_month

  def get_violation_details(self):
    print("Driver ID: ", self.driver_id)
    print("Violation Type: ", self.violation_type)
    print("Fine Amount: ", self.fine_amount)
    print("Speed Over Limit: ", self.speed_over_limit)
    print("Number of Violation: ", self.number_of_violation)
    print("Payment Status: ", self.payment_status)
    print("Violation Month: ", self.violation_month)


class SpeedingViolation(Violation):
  def __init__(self, driver_id,violation_type, fine_amount,speed_over_limit, numbere_of_violation, payment_status, violation_month):
    super().__init__(self, driver_id, "Speeding", fine_amount,speed_over_limit, numbere_of_violation, payment_status, violation_month)




```
